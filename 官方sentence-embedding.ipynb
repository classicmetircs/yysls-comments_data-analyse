```python
import pandas as pd
import re
from datetime import datetime
# 合并官方文本
lists_official = []
pre_official = []
df_bv_cut_info = pd.read_csv('data/bv_info_cleaned.csv', encoding='utf-8')
df_opus_cut_content = pd.read_csv('data/opus_content_cleaned.csv', encoding='utf-8')

lists_official.append(df_bv_cut_info[['内容', '发布时间']].rename(columns={'简介': '内容'}))
lists_official.append(df_opus_cut_content[['内容', '发布时间']].rename(columns={'简介': '内容'}))
df_bv_pre_info = pd.read_csv('data/bv_info_simple.csv', encoding='utf-8')
df_opus_pre_content = pd.read_csv('data/opus_content_simple.csv', encoding='utf-8')
pre_official.append(df_bv_pre_info[['合并文本', '发布时间']].rename(columns={'简介': '内容'}))
pre_official.append(df_opus_pre_content[['合并文本', '发布时间']].rename(columns={'简介': '内容'}))
df_cut_official = pd.concat(lists_official, ignore_index=True)
df_pre_official = pd.concat(pre_official, ignore_index=True)
timestamp_official = df_cut_official['发布时间'].dropna().astype(str).tolist()
emb_input_official = df_pre_official['合并文本'].dropna().astype(str).tolist()
sentences_official = df_cut_official['内容'].dropna().astype(str).tolist()
print(emb_input_official[:2])
print(sentences_official[:2])
# 计算列表的最长文本长度（如果列表为空则返回0）
max_length = max((len(sentence) for sentence in emb_input_official), default=0)
print("检查文本长度是否一致",len(emb_input_official), len(sentences_official))
print("最长文本长度：", max_length)
```

```python
def time_format(timestamp):
    if timestamp is None:
        return ''
    ts = str(timestamp).strip()
    # 删除常见前缀
    ts = re.sub(r'编辑于 ', '', ts)
    # 直接匹配绝对日期（含或不含时分秒）
    try:
        # 优先将常见中文格式解析为 datetime
        dt = pd.to_datetime(ts, errors='coerce', dayfirst=False)
        if not pd.isna(dt):
            return dt.strftime('%Y-%m')
    except Exception:
        pass
    # 最后退回到最简单的替换（保证可读性）
    ts = re.sub(r'年', '-', ts)
    ts = re.sub(r'月', '-', ts)
    ts = re.sub(r'日', '', ts)
    #删除时间，只保留年月
    ts = re.sub(r' \d{1,2}:\d{1,2}(:\d{1,2})?$', '', ts)
    ts = re.sub(r'-\d{1,2}$', '', ts)
    return ts

official_dates = [time_format(ts) for ts in timestamp_official]
print("处理后的发布时间示例：", official_dates[40:])

bv_titles = df_bv_cut_info['标题'].dropna().astype(str).tolist()
opus_titles = df_opus_cut_content['标题'].dropna().astype(str).tolist()
official_titles = bv_titles + opus_titles

#使用正则表达式删除标题里面的bilibili和哔哩哔哩
official_titles = [re.sub(r'(?i)bilibili|哔哩哔哩', '', title).strip() for title in official_titles]

print("标题示例：", official_titles[:5])
```

```python
#保存一个新的文件，包含emb_input_official ,sentences_official和official_datas三列
df_official_final = pd.DataFrame({
    '原始文本': emb_input_official,
    '分词结果': sentences_official,
    '发布时间': official_dates,
    "标题":official_titles
})
df_official_final.to_csv('data/official_lists.csv', index=False, encoding='utf-8')
```

```python
import pandas as pd
import re
# 读取刚才保存的文件
df=pd.read_csv('data/official_lists.csv', encoding='utf-8')
emb_input_official = df['原始文本'].dropna().astype(str).tolist()
#使用正则表达式，将开头“《》”中的内容从开头转移到文末，用“——”连接。正文中也有很多《》符号，但不处理
import re
def move_title_to_end(text):
    pattern = r'^《(.*?)》:'
    match = re.match(pattern, text)
    if match:
        title = match.group(1)
        new_text = re.sub(pattern, '', text).strip()
        #删除文末“_哔哩哔哩_bilibili”
        title = re.sub(r'(?i)bilibili|哔哩哔哩', '', title).strip()
        
        return f"{new_text} —— {title}"
    return text
emb_input_official = [move_title_to_end(text) for text in emb_input_official]
print(emb_input_official[:2])
```

```python
from sentence_transformers import SentenceTransformer
model = SentenceTransformer("./youtu-model",trust_remote_code=True)
documents_embeddings = model.encode_document(emb_input_official, batch_size=32, show_progress_bar=True)
#保存模型
import numpy as np
np.save('data/official_sentences_embeddings.npy', documents_embeddings)
```

```python
from math import dist
import numpy as np
import pandas as pd
from bertopic import BERTopic
from openai import OpenAI
from umap import UMAP
from hdbscan import HDBSCAN
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.feature_extraction.text import CountVectorizer
from bertopic.vectorizers import ClassTfidfTransformer
from sentence_transformers import SentenceTransformer

#读取数据
df_official_final = pd.read_csv('data/official_lists.csv', encoding='utf-8')
sentences_official = df_official_final['分词结果'].dropna().astype(str).tolist()
official_dates = df_official_final['发布时间'].dropna().astype(str).tolist()

#导入向量
#embedding_model = SentenceTransformer('richinfoai/ritrieve_zh_v1')
#embedding_model = SentenceTransformer('Kingsoft-LLM/QZhou-Embedding-Zh')
embedding_model = SentenceTransformer("./youtu-model",trust_remote_code=True)
embeddings = np.load('data/official_sentences_embeddings.npy')
print(type(embeddings), embeddings.shape)

# 2. 创建UMAP降维模型
umap_model = UMAP(
  n_neighbors= 18,
  n_components= 9,
  min_dist=0.3,
  metric='cosine',
  random_state=36
)
reduced_embeddings = umap_model.fit_transform(embeddings)

# 3. 创建HDBSCAN聚类模型
# 如果要建设离群值，可以减小下面两个参数min_cluster_size min_samples
# https://hdbscan.readthedocs.io/en/latest/faq.html
hdbscan_model = HDBSCAN(
  min_cluster_size=6,
  min_samples=1,
  metric='euclidean',
  max_cluster_size=240,
)

# 4. 创建CountVectorizer模型
stpwrd=open("stopwords.txt","r",encoding='utf-8').read()
stop_words= stpwrd.split("\n")
vectorizer_model =  CountVectorizer(stop_words=stop_words)
#vectorizer_model =  TfidfVectorizer(stop_words=stop_words)
ctfidf_model = ClassTfidfTransformer(reduce_frequent_words=True)

# 5. 选择主题表示
from bertopic.representation import MaximalMarginalRelevance
from bertopic.representation import KeyBERTInspired
# The main representation of a topic
aspect_model2 = [KeyBERTInspired(top_n_words=20),MaximalMarginalRelevance(diversity=.8)]
aspect_model= KeyBERTInspired(top_n_words=10)
main_representation  = MaximalMarginalRelevance(diversity=0.3, top_n_words=12)

# Add all models together to be run in a single `fit`
representation_model = {
   "Main": main_representation,
   'Aspect':aspect_model,
   "Aspect2":  aspect_model2 
}

# 6. 创建BERTopic模型
topic_model = BERTopic(
  embedding_model=embedding_model,
  vectorizer_model=vectorizer_model,
  ctfidf_model=ctfidf_model,
  umap_model=umap_model,
  hdbscan_model=hdbscan_model,
  representation_model=representation_model,
  verbose=True
)

# 训练模型
topics, probs = topic_model.fit_transform(sentences_official, embeddings=embeddings)

# 获取聚类结果
topic_info = topic_model.get_topic_info()

#保存聚类结果
topic_docs = topic_model.get_document_info(sentences_official)
topic_docs.to_csv('./聚类结果.csv')

topic_info
```
